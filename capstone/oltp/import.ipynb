```python
import pandas as pd
import mysql.connector

# 1. Load your large CSV file
df = pd.read_csv('/Users/juanzinser/Documents/projects/data-engineering-certificate/capstone/oltp/oltpdata.csv',header=None )

# 2. Connect to your MySQL database
conn = mysql.connector.connect(
    host='localhost',
    user='root',
    password='',
    database='sales'
)
cursor = conn.cursor()

# 3. Stream and insert data fast
query = "INSERT INTO sales_data VALUES (%s, %s, %s, %s, %s)"
for index, row in df.iterrows():
    cursor.execute(query, tuple(row))

conn.commit()
cursor.close()
conn.close()
print("Successfully imported all rows!")
```
